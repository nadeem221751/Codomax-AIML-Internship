```python
# ============================================================
# Linear Regression using Scikit-learn
# Expected Outcome: Model Trained Successfully
# ============================================================

import pandas as pd
from sklearn.linear_model import LinearRegression

# ============================================================
# 1. LOAD DATASET
# ============================================================

df = pd.read_csv("student_scores.csv")

print("========== DATASET ==========")
print(df)

# ============================================================
# 2. SELECT FEATURE AND TARGET
# ============================================================

# Feature (Input)
# We use Math marks to predict the student's Score.

X = df[["Math"]]

# Target (Output)
y = df["Score"]

print("\n========== FEATURE AND TARGET ==========")

print("Feature (X):")
print(X)

print("\nTarget (y):")
print(y)

# ============================================================
# 3. CREATE LINEAR REGRESSION MODEL
# ============================================================

model = LinearRegression()

print("\n========== MODEL ==========")
print("Linear Regression model created successfully.")

# ============================================================
# 4. TRAIN THE MODEL
# ============================================================

model.fit(X, y)

print("\n========== TRAINING ==========")
print("Model trained successfully!")

# ============================================================
# 5. DISPLAY MODEL PARAMETERS
# ============================================================

print("\n========== MODEL PARAMETERS ==========")

print("Coefficient:", model.coef_[0])
print("Intercept:", model.intercept_)

# ============================================================
# 6. DISPLAY LINEAR REGRESSION EQUATION
# ============================================================

print("\n========== REGRESSION EQUATION ==========")

print(
    f"Score = {model.coef_[0]:.2f} * Math + "
    f"{model.intercept_:.2f}"
)

# ============================================================
# 7. TEST THE TRAINED MODEL
# ============================================================

new_student = pd.DataFrame({
    "Math": [90]
})

prediction = model.predict(new_student)

print("\n========== PREDICTION ==========")

print("Math Marks:", new_student["Math"].iloc[0])
print("Predicted Score:", round(prediction[0], 2))

# ============================================================
# 8. FINAL RESULT
# ============================================================

print("\n==============================================")
print(" Linear Regression Model Trained Successfully")
print("==============================================")
```
